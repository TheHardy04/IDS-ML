# Cross Validation

To optimize our model is to perform [Cross Validation](https://en.wikipedia.org/wiki/Cross-validation_(statistics)). It ensure that our model is at its best capacities.

## Importation of dataset, model and all prerequesites

```python
import joblib
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import cross_val_score
```

```python
loaded_pipeline = joblib.load('../jobs/preprocessing.joblib')
loaded_model = joblib.load('../jobs/logistic_regression_PCA_model.joblib')
data = pd.read_csv('../data/rt_iot2022.csv')
```

We perform the OneHot Encoder on categorical features.

```python
X = data.drop('Attack_type', axis=1)
y = data['Attack_type']

categorical_feature = ['proto', 'service']
onehot = OneHotEncoder(sparse_output = False)
encoded_categorical = onehot.fit_transform(data[categorical_feature])

encoded_df = pd.DataFrame(encoded_categorical, columns = onehot.get_feature_names_out(categorical_feature))
X_encoded = X.drop(categorical_feature, axis = 1)
X_encoded = pd.concat([X_encoded, encoded_df], axis = 1)
X_encoded.head()
```

The we compute the remaining of the pre-processing.

```python
X_preprocessed = loaded_pipeline.transform(X_encoded)
```

## [Cross Validation Score](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.cross_val_score.html)

To be sure that we can get a better result with cross validation method we will evaluate 4 scores with cross validation.

```python
for score in ['accuracy', 'precision_weighted', 'recall_weighted', 'f1_weighted']:
    cv_scores = cross_val_score(loaded_model, X_preprocessed, y, cv=5, scoring=score)
    print(f"Cross-validation {score} scores: {cv_scores}")
    print(f"Mean {score}: {cv_scores.mean()}")
```

Comparison of baseline model and the mean model after cross validation : 

|           | Base model | Mean of Cross validation models |
|-----------|------------|---------------------------------|
| Accuracy  | 0.977      | 0.973                           |
| Precision | 0.977      | 0.973                           |
| Recall    | 0.977      | 0.973                           |
| F1-score  | 0.976      | 0.971                           |

As we can see after cross validation we lose about 0.4% to 0.5% in all metrics. So using cross validation might not find a better model, our model is for now the best !
