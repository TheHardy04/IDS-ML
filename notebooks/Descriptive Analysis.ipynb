# Descriptive Analysis - RT-IoT2022

Simple analysis of IoT dataset for intrusion detection.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Data Overview

```python
df = pd.concat([X, y], axis=1)

print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
print(f"\nFirst rows:")
df.head()
```

```python
print("Column types:")
df.dtypes
```

## 2. Descriptive Statistics

```python
df.describe()
```

## 3. Missing Values

```python
print(df.isnull().sum())
if df.isnull().sum().sum() == 0:
    print("Complete dataset")
```

## 4. Class Distribution

```python
target = y.columns[0]
print(df[target].value_counts())
```

```python
df[target].value_counts().plot(kind='bar', figsize=(10, 5))
plt.title('Class Distribution')
plt.xlabel('Class')
plt.ylabel('Count')
plt.show()
```

## 5. Feature Distribution

```python
X.iloc[:, :6].hist(figsize=(12, 8), bins=30)
plt.tight_layout()
plt.show()
```

## 6. Correlation Analysis

```python
X_num = X.select_dtypes(include=['number'])

matrix = X_num.iloc[:, :85].corr()

plt.figure(figsize=(16, 14))
sns.heatmap(matrix, annot=False, cmap='coolwarm', center=0, 
            square=True, linewidths=0.5, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix', fontsize=16, pad=20)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
